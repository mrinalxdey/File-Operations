```python
import glob
import pandas as pd
import os
```

```python
glob.glob('retail_db/**', recursive=True)
```

```python
source_file = glob.glob('retail_db/*/part-*')
```

```python
for file in source_file:
    file_paths = file.split('\\')
    print(file_paths)
```

```python
base_dir = 'retail_db_json'
file = source_file[0]
file_path = file.split('\\')
```

```python
def txt_to_json(source_files, base_dir):
    for file in source_files:
        file_paths = file.split('\\')
        ds_name = file_paths[-2]
        file_name = file_paths[-1].replace('.txt', '.json')

        txt_file = pd.read_csv(file)

        folder_path = f'{base_dir}/{ds_name}'
        json_file_path = f'{folder_path}/{file_name}'

        os.makedirs(folder_path, exist_ok=True)

        txt_file.to_json(
            json_file_path,
            orient='records',
            indent=4
        )

        print(f'Created: {json_file_path}')
```

```python
txt_to_json(source_file, base_dir)
```
