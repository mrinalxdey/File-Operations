# File Handling

```python
# create a file using write
with open('data.txt', 'w') as file:
    content = file.write("Hello World\n")
```

```python
# append
with open('data.txt', 'a') as file:
    file.write('Adding more content')
```

```python
with open('data.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
with open('data.txt', 'r') as file:
    for line in file:
        print(line.strip())
    print('END')
```

```python
import csv
with open('..\\data.csv', 'r') as file:
    reader = csv.reader(file)
    print(reader)
    data = [row for row in reader]
print(data)
```

```python
with open('..\\data.csv', 'r') as file:
    reader = csv.DictReader(file, delimiter='|')
    data = list(reader)
    for row in data:
        print(row)
```

# Python Collections

```python
from collections import namedtuple, defaultdict, Counter
```

```python
Employee = namedtuple('Employee', ['name', 'dept'])
e1 = Employee('Alice', 'IT')
e2 = Employee('Bob', 'Finance')
print(e2)

d = defaultdict(list)
d['IT'].append('Alice')
d['Finance'].append('Bob')
print(d)

c = Counter(['apple', 'banana', 'apple', 'apple'])
print(c['apple'])
```

# Processing Python List

```python
data = [1,2,3,4,5]

even_squares = [x*x for x in data if x % 2 == 0]
print(even_squares)
```

```python
total = sum(data)
print(total)
average = sum(data)/len(data)
print(round(average))
```

# Lambda Functions

```python
square = lambda x: x**2
print(square(5))
```

```python
nums = [10, 15, 20, 25, 30]
squares = list(map(lambda x: x**2, nums))
print(squares)   # Output: [10, 20, 30]
```

```python
nums = [10, 15, 20, 25, 30]
evens = list(filter(lambda x: x % 2 == 0, nums))
odds = list(filter(lambda x: x%2 != 0, nums))
print(evens) 
print(odds)
```

# Filtering Data

```python
data = [100,200,300,400]

filtered = list(filter(lambda x: x > 150, data))
print(filtered)
```
