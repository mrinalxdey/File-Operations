# File Handling

```python
# create a file using write
with open('data.txt', 'w') as file:
    content = file.write("Hello World\n")
```

```python
# append
with open('data.txt', 'a') as file:
    file.write('Adding more content')
```

```python
with open('data.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
with open('data.txt', 'r') as file:
    for line in file:
        print(line.strip())
    print('END')
```

```python
import csv
with open('..\\data.csv', 'r') as file:
    reader = csv.reader(file)
    print(reader)
    data = [row for row in reader]
print(data)
```

```python
with open('..\\data.csv', 'r') as file:
    reader = csv.DictReader(file, delimiter='|')
    data = list(reader)
    for row in data:
        print(row)
```

# Python Collections

```python
from collections import namedtuple, defaultdict, Counter
```

```python
Employee = namedtuple('Employee', ['name', 'dept'])
e1 = Employee('Alice', 'IT')
e2 = Employee('Bob', 'Finance')
print(e2)

d = defaultdict(list)
d['IT'].append('Alice')
d['Finance'].append('Bob')
print(d)

c = Counter(['apple', 'banana', 'apple', 'apple'])
print(c['apple'])
```

# Processing Python List

```python
data = [1,2,3,4,5]

even_squares = [x*x for x in data if x % 2 == 0]
print(even_squares)
```

```python
total = sum(data)
print(total)
average = sum(data)/len(data)
print(round(average))
```

# Lambda Functions

```python
square = lambda x: x**2
print(square(5))
```

```python
nums = [10, 15, 20, 25, 30]
squares = list(map(lambda x: x**2, nums))
print(squares)   # Output: [10, 20, 30]
```

```python
nums = [10, 15, 20, 25, 30]
evens = list(filter(lambda x: x % 2 == 0, nums))
odds = list(filter(lambda x: x%2 != 0, nums))
print(evens) 
print(odds)
```

# Filtering Data

```python
data = [100,200,300,400]

filtered = list(filter(lambda x: x > 150, data))
print(filtered)
```

```python
data = [100, -1, 2000, None, 0]

cleaned_data = list(filter(lambda x: x != None and x > 0, data))
print(cleaned_data)
```

```python
employees = [
    {
        "name": "Alice",
        "salary": 90000
    },
    {
        "name": "Bob",
        "salary": 75000
    },
    {
        "name": "Eve",
        "salary": 400000
    }
]
high_earners = list(filter(lambda emp: emp['salary'] > 80000, employees))
print(high_earners)
```

```python
name = ['Yashi', 'Alice', 'Bob', 'Yashi', 'Bob', 'Isha']
unique_name = list(set(name))
print(name)
print(unique_name)
```

```python
name = {'Yashi', 'Alice', 'Bob'}
name1 = {'Yashi', 'Bob'}

non_common = name - name1
print(non_common)
```

```python
employees = [
    {
        "name": "Alice",
        "dept": 'IT'
    },
    {
        "name": "Bob",
        "dept": 'HR'
    },
    {
        "name": "Eve",
        "dept": 'IT'
    }
]
unique_values = set(map(lambda x: x['dept'], employees))
print(unique_values)
```

# Sort python list using key

```python
people = [
    {
        "name": "Alice",
        "age": 44
    },
    {
        "name": "Bob",
        "age": 30
    }
]
sorted_people = sorted(people, key = lambda x: x['age'])
print(sorted_people)
```

```python
names = ['Tom', 'Alice', 'Bob']
sorted_names = sorted(names, key = len)
sorted_names
```
